```python
import tensorflow as tf
import time

print(f"TensorFlow versão: {tf.__version__}")
print("Dispositivos disponíveis:", tf.config.list_physical_devices())

x = tf.random.normal([2048, 2048])
y = tf.random.normal([2048, 2048])

start = time.perf_counter()
z = tf.matmul(x, y)
end = time.perf_counter()

print(f"Multiplicação de matrizes (CPU): {end - start:.4f} segundos")

```

```python
import onnxruntime as ort
import numpy as np
import time

print("Providers disponíveis:", ort.get_available_providers())

session = ort.InferenceSession("mobilenetv2.onnx", providers=["DmlExecutionProvider"])
input_name = session.get_inputs()[0].name
input_data = np.random.rand(1, 3, 224, 224).astype(np.float32)

start = time.perf_counter()
output = session.run(None, {input_name: input_data})
end = time.perf_counter()

print(f"Inferência ONNX (NPU): {end - start:.4f} segundos")
```

```python
import torch
from torch_directml import device as dml_device
import time

device = dml_device()
print(f"Usando GPU DirectML: {device}")

x = torch.randn(2048, 2048, device=device)
y = torch.randn(2048, 2048, device=device)

start = time.perf_counter()
z = x @ y
end = time.perf_counter()

print(f"Multiplicação de matrizes (GPU): {end - start:.4f} segundos")
```
