## Importing time series and pandas libraries

```python
# Import with pandas
import pandas as pd
# Let's read the CSV file into a pandas DataFrame
df = pd.read_csv('BTC_USD_2013-10-01_2021-05-18-CoinDesk.csv', parse_dates=['Date'], index_col='Date') # parse the 'Date' column as datetime and set it as the index
# Display the first few rows of the DataFrame
print(df.head())
```

We've collectedted a large dataset of Bitcoin prices over time (8 Years). 

Typically deep learning models usually like lots and lots and lots of samples (where lots and lots and lots can thousands to tens of thousands to millions).
A smaller number of samples is something you'll often run into with time series data problems.

> Note: The **seasonality** of a time series dataset is the number of samples per year. For Bitcoin data, seasonality is daily (365) because one sample is collected per day, giving 365 samples per year.
